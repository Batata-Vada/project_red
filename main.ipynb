```python
df = df.drop(columns=["id", "external_id", "url"])
```

```python
OHE = OneHotEncoder(sparse_output=False, handle_unknown='error')

encoded = OHE.fit_transform(df[["platform"]])

encoded_df = pd.DataFrame(encoded, columns=OHE.get_feature_names_out(["platform"]), index=df.index)

df = pd.concat([df.drop(columns="platform"),encoded_df], axis=1)
```

```python
df.head()
```

```python
df["title length"] = df["title"].apply(lambda x: len(x))
df = df.drop(columns=["title"])
```
