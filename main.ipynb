```python
import torch
import pandas as pd 
import sqlite3
import json
```

```python
conn = sqlite3.connect("app.db")
```

```python
tabels = pd.read_sql_query("SELECT name FROM sqlite_master WHERE type='table';", conn)
tabels
```

```python
df_01 = pd.read_sql_query("SELECT *" \
                          "FROM user", conn)
df_01
```

```python
df_02 = pd.read_sql_query("SELECT *" \
                          "FROM session", conn)
df_02
```

```python
df_03 = pd.read_sql_query("SELECT *" \
                          "FROM problems", conn)
df_03.head()
```

```python
sum = df_03.groupby("platform").size()
sum
```

```python
df_03["difficulty_raw"] = df_03["difficulty_raw"].apply(json.loads)
df_03["tags_raw"] = df_03["tags_raw"].apply(json.loads)
df_03["stats_raw"] = df_03["stats_raw"].apply(json.loads)
df_03["metadata_raw"] = df_03["metadata_raw"].apply(json.loads)
```

```python
df_03.to_csv("app_00.csv", index=False)
```
