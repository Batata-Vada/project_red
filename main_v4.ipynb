```python
import onnxmltools
from onnxmltools.convert.common.data_types import FloatTensorType
from xgboost import XGBRegressor
from sklearn.multioutput import MultiOutputRegressor
from sklearn.metrics import mean_squared_error
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
import joblib
```

```python
df = pd.read_csv("project_red_center_width.csv")
df.head()
```

```python
features = ["user_rating", "last_problem_rating", "mean_solve_time_minutes", "time_std_minutes", "problems_solved"]

targets = ["center", "width"]

X = df[features]
Y = df[targets]
```

```python
X_train, X_test, Y_train, Y_test = train_test_split(X, Y, test_size=0.2, random_state=11037)
```

```python
scaler_x = StandardScaler()

X_train = scaler_x.fit_transform(X_train)
X_test =  scaler_x.fit_transform(X_test)
```

```python
joblib.dump(scaler_x, "scaler_x_v4.pkl")
```

```python
scaler_y = StandardScaler()

Y_train = scaler_y.fit_transform(Y_train)
Y_test =  scaler_y.fit_transform(Y_test)
```

```python
joblib.dump(scaler_y, "scaler_y_v4.pkl")
```

```python
XGB = XGBRegressor(n_estimators=500, max_depth=4, learning_rate=0.05, objective="reg:squarederror", random_state=11037, n_jobs=-1)
```

```python
model = MultiOutputRegressor(XGB)
```

```python
model.fit(X_train, Y_train)
```

```python
prd = model.predict(X_test)
```

```python
center_MSE = mean_squared_error(Y_test[:,0], prd[:,0])
center_MSE
```

```python
width_MSE = mean_squared_error(Y_test[:,1], prd[:,1])
width_MSE
```

```python
MSE = mean_squared_error(Y_test, prd)
MSE
```
